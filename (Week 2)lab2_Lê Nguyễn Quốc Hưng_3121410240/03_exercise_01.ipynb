```python
import math

# Nhập số thực x
x = float(input("Nhập giá trị x: "))

# Tính f(x)
if x == 0:
    print("Biểu thức không xác định vì có chia cho 0.")
else:
    f = x + (x**5 / math.factorial(5)) + (math.sqrt(abs(x)) / (x**(3/2)))
    print(f"Giá trị f({x}) = {f}")
```

```python
# Prace 3
# Nhập hệ số a và b
a = float(input("Mời bạn nhập hệ số a: "))
b = float(input("Mời bạn nhập hệ số b: "))

# Xử lý các trường hợp
if a == 0:
    if b == 0:
        print("Phương trình 0x + 0 = 0 có vô số nghiệm.")
    else:
        print(f"Phương trình 0x + {b} = 0 vô nghiệm.")
else:
    x = -b / a
    print(f"Phương trình {a}x + {b} = 0 có 1 nghiệm x = {x:.2f}")
```

```python
# Prace 4
# Từ điển ánh xạ tiếng Anh -> (số, tiếng Việt)
mapping = {
    "zero": ("0", "không"),
    "one": ("1", "một"),
    "two": ("2", "hai"),
    "three": ("3", "ba"),
    "four": ("4", "bốn"),
    "five": ("5", "năm"),
    "six": ("6", "sáu"),
    "seven": ("7", "bảy"),
    "eight": ("8", "tám"),
    "nine": ("9", "chín")
}

# Nhập chuỗi
s = input("Mời bạn nhập chuỗi ký số s: ").strip().lower()

# Kiểm tra hợp lệ
if s in mapping:
    so, tiengviet = mapping[s]
    print(f'Chuỗi vừa nhập hợp lệ!')
    print(f'"{s.capitalize()}" biểu diễn số cho "{so}" và ứng với tiếng Việt "{tiengviet}".')
else:
    print("Chuỗi vừa nhập không hợp lệ!")
```

```python
# Practice 6
# Hàm kiểm tra số nguyên tố
def is_prime(n: int) -> bool:
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):  # chỉ cần duyệt tới căn bậc 2 của n
        if n % i == 0:
            return False
    return True

# Nhập số nguyên dương n
n = int(input("Mời bạn nhập số nguyên dương n: "))

# Kiểm tra và in kết quả
if is_prime(n):
    print(f"{n} là số nguyên tố.")
else:
    print(f"{n} không phải là số nguyên tố.")
```

```python
# Practice 7
import math

# Hàm giải phương trình bậc 2
def giai_pt_bac2(a, b, c):
    if a == 0:  # Trường hợp trở thành phương trình bậc 1
        if b == 0:
            if c == 0:
                return "Phương trình có vô số nghiệm."
            else:
                return "Phương trình vô nghiệm."
        else:
            x = -c / b
            return f"Phương trình bậc 1 {b}x + {c} = 0 có nghiệm x = {x:.2f}"
    else:
        delta = b**2 - 4*a*c
        if delta < 0:
            return f"Phương trình bậc 2 {a}x^2 + {b}x + {c} = 0 vô nghiệm."
        elif delta == 0:
            x = -b / (2*a)
            return f"Phương trình bậc 2 {a}x^2 + {b}x + {c} = 0 có nghiệm kép x1 = x2 = {x:.2f}"
        else:
            x1 = (-b + math.sqrt(delta)) / (2*a)
            x2 = (-b - math.sqrt(delta)) / (2*a)
            return f"Phương trình bậc 2 {a}x^2 + {b}x + {c} = 0 có 2 nghiệm x1 = {x1:.2f}, x2 = {x2:.2f}"

# Nhập dữ liệu có kiểm tra
while True:
    try:
        values = input("Mời bạn nhập hệ số a, b, c: ").split()
        if len(values) != 3:
            raise ValueError("Phải nhập đúng 3 số!")
        a, b, c = map(float, values)
        break
    except ValueError:
        print("Bạn nhập sai! Mời bạn nhập lại!")

# Gọi hàm và in kết quả
print(giai_pt_bac2(a, b, c))
```

```python
# Practice 8
import random

# Hàm nhập mảng số từ bàn phím
def NhapMang():
    while True:
        try:
            arr = list(map(int, input("Nhập mảng số (cách nhau bởi khoảng trắng): ").split()))
            return arr
        except ValueError:
            print("Bạn nhập sai! Mời nhập lại (chỉ nhập số nguyên).")

# Hàm xuất mảng
def XuatMang(arr):
    print("Mảng:", arr)

# Hàm sinh mảng số ngẫu nhiên
def SinhNgauNhien(n, a, b):
    arr = [random.randint(a, b) for _ in range(n)]
    return arr

# Hàm tính tổng, đếm số chẵn và lẻ
def TinhTongChanLe(arr):
    tong = sum(arr)
    chan = sum(1 for x in arr if x % 2 == 0)
    le = len(arr) - chan
    return tong, chan, le

# Hàm tách mảng thành dãy chẵn và dãy lẻ
def TimChanLe(arr):
    a_chan = [x for x in arr if x % 2 == 0]
    a_le = [x for x in arr if x % 2 != 0]
    return a_chan, a_le


# =============================
# Ví dụ chạy thử
# =============================
a = NhapMang()
XuatMang(a)

b = SinhNgauNhien(10, -10, 10)
XuatMang(b)

tong, chan, le = TinhTongChanLe(b)
print(f"Tổng = {tong}, số phần tử chẵn = {chan}, số phần tử lẻ = {le}")

a_chan, a_le = TimChanLe(b)
print("Dãy chẵn:", a_chan)
print("Dãy lẻ:", a_le)
```

```python
# practice 9

# Nhập dãy ký tự cách nhau bởi khoảng trắng
a = input("Nhập dãy ký tự (cách nhau bởi khoảng trắng): ").split()

# 1. In các ký tự không trùng, sắp xếp tăng dần
unique_sorted = sorted(set(a))
print("Các ký tự không trùng (tăng dần):", unique_sorted)

# 2. Đếm số lần xuất hiện của từng ký tự
count_dict = {}
for ch in a:
    count_dict[ch] = count_dict.get(ch, 0) + 1

print("Số lần xuất hiện của các ký tự:")
for k, v in count_dict.items():
    print(f"'{k}': {v}")
```
